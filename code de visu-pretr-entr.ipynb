```python
# Étape 6 : Gestion des valeurs aberrantes
```

```python
# Justification : Les boxplots montrent des valeurs aberrantes significatives dans `sbytes` et `dbytes`. On applique un filtre pour limiter leur impact.
```

```python
if 'sbytes' in df.columns and 'dbytes' in df.columns:
    for col in ["sbytes", "dbytes"]:
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1
        df = df[(df[col] >= Q1 - 1.5 * IQR) & (df[col] <= Q3 + 1.5 * IQR)]
else:
    print("Les colonnes 'sbytes' ou 'dbytes' sont absentes.")
```

```python
# Étape 7 : Standardisation des colonnes
```

```python
# Justification : Les modèles comme les SVM sont sensibles aux échelles des données. Les colonnes sont standardisées pour de meilleures performances.
```

```python
cols_to_normalize = ["sbytes", "dbytes", "Sload", "Dload", "Spkts", "Dpkts"]
std_scaler = StandardScaler()
df[cols_to_normalize] = std_scaler.fit_transform(df[cols_to_normalize])
```

```python
from imblearn.over_sampling import SMOTE

# Vérification de la colonne cible pour équilibrage
if "Label" in df.columns or "label" in df.columns:  # Adapté pour `label` ou `Label`
    target_col = "Label" if "Label" in df.columns else "label"
    
    # Séparation des caractéristiques et de la cible
    X = df.drop(columns=[target_col])
    y = df[target_col]
    
    # Application de SMOTE
    smote = SMOTE(random_state=42)
    X_resampled, y_resampled = smote.fit_resample(X, y)
    
    # Fusion des caractéristiques et de la cible resamplées
    df = pd.concat([X_resampled, y_resampled], axis=1)
    print("Les classes ont été équilibrées avec succès.")
else:
    print("La colonne cible pour l'équilibrage des classes est absente.")
```

```python
import pandas as pd
import numpy as np
from sklearn.tree import DecisionTreeClassifier
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import (
    classification_report,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)
from sklearn.model_selection import train_test_split

# 1. Vérification des types de données
print("Types de données dans X:")
print(X.dtypes)

# 2. Conversion des colonnes non-numériques
for col in X.select_dtypes(include=['object']).columns:
    if X[col].nunique() < 10:
        le = LabelEncoder()
        X[col] = le.fit_transform(X[col])
    else:
        print(f"Colonne '{col}' contient des valeurs complexes - à traiter manuellement")
        X = X.drop(col, axis=1)

# 3. Vérification
assert not any(X.dtypes == 'object'), "Il reste des colonnes non-numériques!"

# 4. Split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# 5. Entraînement
clf = DecisionTreeClassifier(random_state=42)
clf.fit(X_train, y_train)

# 6. Prédiction
y_pred = clf.predict(X_test)
y_proba = clf.predict_proba(X_test)[:, 1]

# 7. Métriques globales
print("\n=== Métriques globales ===")
print(f"Accuracy : {accuracy_score(y_test, y_pred):.4f}")
print(f"Precision : {precision_score(y_test, y_pred):.4f}")
print(f"Recall : {recall_score(y_test, y_pred):.4f}")
print(f"F1-score : {f1_score(y_test, y_pred):.4f}")
print(f"AUC-ROC : {roc_auc_score(y_test, y_proba):.4f}")
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_curve, auc

# =============== ÉTAPE 1 : Préparation ===============
# ⚠️ Remplace ces deux lignes par ton propre chargement des données
# X = df.drop('Label', axis=1)
# y = df['Label']

# 1. Supprimer les colonnes à haute cardinalité ou très peu utiles
threshold_cardinality = 100  # ne pas encoder si plus de 100 valeurs uniques
columns_to_drop = []

for col in X.columns:
    if X[col].nunique() > threshold_cardinality:
        columns_to_drop.append(col)
    elif X[col].isnull().mean() > 0.5:
        columns_to_drop.append(col)
    elif X[col].nunique() <= 1:
        columns_to_drop.append(col)

X_reduced = X.drop(columns=columns_to_drop)

# 2. Encodage LabelEncoder pour les colonnes catégorielles restantes
for col in X_reduced.select_dtypes(include=['object', 'category']).columns:
    le = LabelEncoder()
    try:
        X_reduced[col] = le.fit_transform(X_reduced[col].astype(str))
    except:
        X_reduced = X_reduced.drop(columns=[col])  # au cas où encodage échoue

# 3. Imputation des NaNs
X_reduced = X_reduced.fillna(X_reduced.mean(numeric_only=True))

# =============== ÉTAPE 2 : Réduction des dimensions ===============
# Réduire à 500 colonnes max si trop volumineux
if X_reduced.shape[1] > 500:
    # garder les 500 colonnes les plus variables
    variances = X_reduced.var().sort_values(ascending=False)
    top_cols = variances.head(500).index
    X_reduced = X_reduced[top_cols]

# =============== ÉTAPE 3 : Normalisation et split ===============
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_reduced)

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.3, random_state=42
)

# =============== ÉTAPE 4 : Modèle Naive Bayes ===============
model = GaussianNB()
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
y_proba = model.predict_proba(X_test)

# =============== ÉTAPE 5 : Évaluation ===============
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred, zero_division=0)
recall = recall_score(y_test, y_pred, zero_division=0)
f1 = f1_score(y_test, y_pred, zero_division=0)
fpr, tpr, _ = roc_curve(y_test, y_proba[:, 1])
roc_auc = auc(fpr, tpr)

print("=== Résultats ===")
print(f"Accuracy  : {accuracy:.4f}")
print(f"Precision : {precision:.4f}")
print(f"Recall    : {recall:.4f}")
print(f"F1-score  : {f1:.4f}")
print(f"AUC-ROC   : {roc_auc:.4f}")
```

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, classification_report

# ========== 1. Chargement des données ==========
df = pd.read_csv("UNSW_NB15_combined.csv")

# ========== 2. Suppression des colonnes inutiles ==========
y = df['Label']
X = df.drop(columns=['Label'])

# Suppression des colonnes avec trop de valeurs manquantes ou de haute cardinalité
threshold_cardinality = 100
columns_to_drop = []

object_cols = X.select_dtypes(include=['object', 'category']).columns
for col in object_cols:
    if X[col].nunique() > threshold_cardinality:
        columns_to_drop.append(col)

for col in X.columns:
    if X[col].isnull().mean() > 0.5:
        columns_to_drop.append(col)

X_cleaned = X.drop(columns=set(columns_to_drop))

# ========== 3. Encodage mixte ==========
categorical_cols = X_cleaned.select_dtypes(include=['object', 'category']).columns

le = LabelEncoder()
X_encoded = X_cleaned.copy()

for col in categorical_cols:
    if X_encoded[col].nunique() > 5:
        X_encoded[col] = le.fit_transform(X_encoded[col])

# ========== 4. Normalisation ==========
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_encoded)

# ========== 5. Validation croisée avec Naive Bayes ==========
nb = GaussianNB()

# Utilisation de la validation croisée
cv_scores = cross_val_score(nb, X_scaled, y, cv=5)  # 5-fold cross-validation

# Affichage des résultats
print(f"Cross-validation scores: {cv_scores}")
print(f"Mean CV score: {cv_scores.mean()}")

# ========== 6. Entraînement et évaluation sur l'ensemble de test ==========
# Séparation de l'ensemble d'entraînement et de test pour évaluer de manière standard
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.3, random_state=42)

# Entraînement sur les données d'entraînement
nb.fit(X_train, y_train)

# Prédictions et évaluation
y_pred = nb.predict(X_test)

# Calcul de la précision et du rapport de classification
accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy: {accuracy}")
print("Classification Report:")
print(classification_report(y_test, y_pred))


```

```python
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
from sklearn.naive_bayes import GaussianNB
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
import numpy as np

# Initialisation du modèle et du scaler
pipeline = make_pipeline(StandardScaler(), GaussianNB())

# Initialisation de la validation croisée stratifiée
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

# Listes pour stocker les métriques
accuracy_list = []
precision_list = []
recall_list = []
f1_list = []

# Validation croisée manuelle
for train_index, test_index in cv.split(X, y):
    X_train, X_test = X.iloc[train_index], X.iloc[test_index]
    y_train, y_test = y.iloc[train_index], y.iloc[test_index]

    # Entraînement
    pipeline.fit(X_train, y_train)

    # Prédiction
    y_pred = pipeline.predict(X_test)

    # Calcul des métriques
    accuracy_list.append(accuracy_score(y_test, y_pred))
    precision_list.append(precision_score(y_test, y_pred, average='binary'))
    recall_list.append(recall_score(y_test, y_pred, average='binary'))
    f1_list.append(f1_score(y_test, y_pred, average='binary'))

# Affichage des résultats moyens
print("=== Résultats de la validation croisée (5 folds) ===")
print(f"Accuracy moyenne : {np.mean(accuracy_list):.4f}")
print(f"Precision moyenne : {np.mean(precision_list):.4f}")
print(f"Recall moyen : {np.mean(recall_list):.4f}")
print(f"F1-score moyen : {np.mean(f1_list):.4f}")
```
